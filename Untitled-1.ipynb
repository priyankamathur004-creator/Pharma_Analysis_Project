```python
import ast
def extract_name(val):
    if isinstance(val,str):
        try:
            val = ast.literal_eval(val)
        except(ValueError,SyntaxError):
            return ""
    if isinstance(val,list):
        names = [item.get('name') for item in val if isinstance (item,dict) and 'name' in item]
        return ",".join(names)
    return ""

Pharma_Df['Ingredients_name'] = Pharma_Df['active_ingredients'].apply(extract_name)
Pharma_Df['Ingredients_name'].head()
```

```python
#• Segment based on number of ingredients 
Pharma_Df['num_active_ingredients'] 
Pharma_Df['ingredients_segment'] = pd.cut(Pharma_Df['num_active_ingredients'],bins = [1,2,3,Pharma_Df['num_active_ingredients'],max()],
labels = ['Single','Double','Multi'])
```

```python
# Find maximum number of active ingredients
max_ing = Pharma_Df['num_active_ingredients'].max()

# Define bins without duplicates
Pharma_Df['ingredients_segment'] = pd.cut(
    Pharma_Df['num_active_ingredients'],
    bins=[0,1,2,max_ing+1],   # unique edges only
    labels=['Single','Double','Multi'],
    include_lowest=True
)

Pharma_Df[['num_active_ingredients','ingredients_segment']].head()
```

```python
Pharma_Df['ingredients_segment'] = Pharma_Df['num_active_ingredients'].map(lambda x:'Single' if x==1 else 'Double' if x==2 else 'Multi')
Pharma_Df[['num_active_ingredients','ingredients_segment']].head(10)
```

```python
##• Segment based on dosage form 

mapping = {
    'Tablet':'Solid Oral',
    'Capsule':'Solid Oral',
    'Powder':'Solid Oral',
    'Syrup':'Liquid Oral',
    'Suspension':'Liquid Oral',
    'Drops':'Liquid Oral',
    'Injection':'Parenteral',
    'Cream':'Topical',
    'Ointment':'Topical',
    'Gel':'Topical',
    'Spray':'Respiratory',
    'Inhaler':'Respiratory',
    'Respules':'Respiratory',
    'Solution':'Other',
    'Other':'Other'
}

Pharma_Df['Dosage_form_segment'] = Pharma_Df['Dosage_form'].map(mapping)
Pharma_Df['Dosage_form_segment']
```
